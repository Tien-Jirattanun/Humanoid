```python
import matplotlib.pyplot as plt
import numpy as np

# Load home trajectory data
loaded_home_data = np.load('/home/tien/Documents/GitHub/Humanoid/src/hanuman_control/include/trajectory/q_home_data.npz')
print("Available keys in loaded_home_data:", loaded_home_data.files)

# Extract the home trajectory array
q_home = loaded_home_data['q']  # shape: (num_points, 12)

# Plot each joint trajectory
for i in range(q_home.shape[1]):  # 12 joints
    plt.plot(q_home[:, i], label=f'Joint {i+1}')
plt.xlabel('Trajectory Step')
plt.ylabel('Joint Position')
plt.title('Home Trajectory for All Joints')
plt.legend()
plt.grid()
plt.show()
```
